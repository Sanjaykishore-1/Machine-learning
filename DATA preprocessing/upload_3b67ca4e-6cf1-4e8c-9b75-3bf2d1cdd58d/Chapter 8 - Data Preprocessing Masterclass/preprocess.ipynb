```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import AdaBoostClassifier
```

```python
environment= pd.read_csv("environmental_data.csv")
environment
```

```python
environment.dtypes
```

```python
environment.isna().sum()
```

```python
environment.describe()
```

```python
environment.drop(columns=["Unnamed: 0","Temp C"], inplace=True)
```

```python
## we have removed two colums from the dataset. The first one is an index column and the second one is a temperature column which has a lot of missing values. We will see how to handle missing values in the next section.
```

```python
environment
```

```python
del environment["Ozone"]
```

```python
environment[environment["Solar.R"].isna()]
```

```python
environment.groupby("Weather")["Solar.R"].mean()
```

```python
environment["Solar.R"] = environment["Solar.R"].transform(lambda x: x.fillna(x.mean()).round(2))
```

```python
environment.isna().sum()
```

```python
environment.dropna(inplace=True)
```

```python
environment.dtypes
```

```python
environment["Month"].unique()
```

```python
environment["Month"] = environment["Month"].replace(to_replace = "May", value = '5')
```

```python
environment["Month"] = environment["Month"].astype("int64")
```

```python
environment["Weather"].unique()
```

```python
en_copy1 = environment.copy()
```

```python
en_copy2 = environment.copy()
```

```python
en_copy1
```

```python
label_encoder = LabelEncoder()
```

```python
en_copy1["Weather"] = label_encoder.fit_transform(en_copy1["Weather"])
```

```python
## after the label encoding, the weather column has been transformed into numerical values. The unique values in the weather column have been replaced with integers. 
#The mapping of the original values to the integers can be obtained using the classes_ attribute of the label encoder.
## the weather column the data defined as follows:
# 2: "Sunny"
# 0: "Cloudy"  
# 1: "partly cloudy"
```

```python
en_copy1
```

```python
from sklearn.preprocessing import OneHotEncoder
```
