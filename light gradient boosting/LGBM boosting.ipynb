```python
## data preprocessing & preparing the data for model building
```

```python
del credit_card_data['ID']
```

```python
LE = LabelEncoder()
```

```python
credit_card_data["SEX"] = LE.fit_transform(credit_card_data["SEX"])
credit_card_data["EDUCATION"] = LE.fit_transform(credit_card_data["EDUCATION"])
credit_card_data["MARRIAGE"] = LE.fit_transform(credit_card_data["MARRIAGE"])
```

```python
credit_card_data
```

```python
X = credit_card_data.drop(['DEFAULT'], axis=1)
X
```

```python
y = credit_card_data['DEFAULT']
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=150,stratify=y,shuffle= True)
```

```python
## model building
```

```python
abm = AdaBoostClassifier(n_estimators = 50,learning_rate = 0.3)
gbm = GradientBoostingClassifier(n_estimators = 100,max_depth= 5,learning_rate = 0.3)
xgbm = xgb.XGBClassifier(n_estimators = 100,max_depth= 5,learning_rate = 0.3)
lgbm = lgb.LGBMClassifier(n_estimators = 100,max_depth= 5,learning_rate = 0.3)
```

```python
## training the models
```

```python
%%time
lgbm.fit(X_train, y_train,)
```

```python
%%time
abm.fit(X_train, y_train)
```

```python
%%time
gbm.fit(X_train, y_train)
```

```python
%%time
xgbm.fit(X_train, y_train)
```

```python
y_pred_abm = abm.predict(X_train)
y_pred_gbm = gbm.predict(X_train)
y_pred_xgbm = xgbm.predict(X_train)
y_pred_lgbm = lgbm.predict(X_train)
```

```python
## model testing 
```

```python
# use trained models to predict on test data
# do not refit on X_test/y_test, otherwise test evaluation is invalid

y_pred_abm_test = abm.predict(X_test)
y_pred_gbm_test = gbm.predict(X_test)
y_pred_xgbm_test = xgbm.predict(X_test)
y_pred_lgbm_test = lgbm.predict(X_test)
```
